```python
#selecting the features (x)
#rollnumber and name are not used for predition
X=fig[["CGPA",	"Skills",	"Projects"	,"Internships",	"Certifications",	"DSA_Score"]]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)
```

```python
print("training students:",len(X_train))
print("training students:",len(X_test))
```

```python
#create the machine learning model
model=DecisionTreeClassifier(random_state=42)
```

```python
#train the model
model.fit(X_train,Y_train)
```

```python
print("\nmodel training completed")
```

```python
# test the model
predictions = model.predict(X_test)

print("\nActual values:   ", list(Y_test))
print("Predicted values:", list(predictions))
```

```python
#Accuracy testing
accuracy=accuracy_score(Y_test,predictions)
```
